```python
%load_ext autoreload
%autoreload 2

import numpy as np
import matplotlib.pyplot as plt
import utils as ut

from sklearn.datasets import fetch_openml
```

In this notebook, we can input our choice of parameters and generate the dataset of pairs (structure factor heatmap, MNIST digit label).

```python
# PARAMETERS

T = 2.0 # critical temperature is at 2.27
alpha_over_J = 1.0

sweeps = 10000 # number of sweeps until equilibrium
n_samples = 100 # number of structure factor measurements to average over
interval = 10 # number of sweeps between measurements of structure factor

n_images = 70000 # number of MNIST images to convert
```

```python
# retrieve MNIST dataset via OpenML
X, y = fetch_openml("mnist_784", version=1, return_X_y=True, as_frame=False)
y = y.astype(int)

# normalise to values between 0 and 1
X = X.astype(np.float64) / 255.0
```

```python
X_ising = np.zeros((n_images, 784))

for i in range(n_images):
    X_ising[i] = ut.ising_features(X[i], T, alpha_over_J, sweeps=sweeps, n_samples=n_samples, interval=interval)

    if (i + 1) % 1000 == 0:
        print(f"Processed {i+1}/{n_images} images")
```

```python
# save generated features

np.savez_compressed("ising_features.npz", X=X_ising, y=y, T=T, alpha_over_J=alpha_over_J)
```
