```python
%load_ext autoreload
%autoreload 2

import numpy as np
import matplotlib.pyplot as plt
import utils as ut
```

Here we train one model and observe convergence, similar to `logistic_regression/introduction.ipynb`.

```python
# PARAMETERS

learning_rate = 0.1
epochs = 2000
lam = 0
```

```python
# retrieve stored dataset
X, y, T, alpha_over_J = np.load("ising_features.npz")

X_train = X[:60000]
X_test = X[60000:]

y_train = y[:60000]
y_test = y[60000:]

X_train, X_test = ut.normalise(X_train, X_test)
```

```python
# execute training

W, b, losses = ut.train(X_train, y_train, learning_rate=learning_rate, epochs=epochs, lam=lam, return_losses=True)

plt.plot(losses)
plt.xlabel("Epoch")
plt.ylabel("Cross-entropy loss")
plt.grid(True)
plt.show()
```

```python
# evaluate classifier performance

train_predictions = ut.predict(X_train, W, b)
test_predictions = ut.predict(X_test, W, b)

train_accuracy = np.mean(train_predictions == y_train)
test_accuracy = np.mean(test_predictions == y_test)

print(f"Training accuracy: {train_accuracy:.2%}")
print(f"Test accuracy: {test_accuracy:.2%}")
```
